```python

import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np
import warnings

warnings.filterwarnings('ignore')
```

```python
# Setup visualisasi
sns.set_theme(style="whitegrid")
plt.rcParams.update({
    'font.size': 11,
    'axes.spines.top': False,
    'axes.spines.right': False
})
```

```python
# Load dataset
path = r"D:\Karir\Bootcamp\Data Scientist Rakamin\Week 3\Final Task\Code\train_full_features.csv"
df = pd.read_csv(path)
```

```python
# 1. Target Distribution
plt.figure(figsize=(7, 7))
target_counts = df['TARGET'].value_counts()
plt.pie(
    target_counts, 
    labels=['Lancar', 'Macet'], 
    autopct='%1.1f%%', 
    startangle=140, 
    colors=['#2ecc71', '#e74c3c'], 
    explode=(0, 0.1)
)
plt.title("Proyeksi Rasio Kolektibilitas Nasabah", pad=20)
plt.show()
```

```python
# 2. External Score Distribution (KDE)
plt.figure(figsize=(10, 5))
sns.kdeplot(df[df['TARGET'] == 0]['EXT_SOURCE_MEAN'], label='Lancar', fill=True, color='#3498db')
sns.kdeplot(df[df['TARGET'] == 1]['EXT_SOURCE_MEAN'], label='Macet', fill=True, color='#e74c3c')
plt.title("Distribusi EXT_SOURCE_MEAN terhadap Probabilitas Default")
plt.xlabel("Mean Score")
plt.ylabel("Density")
plt.legend()
plt.tight_layout()
plt.show()
```

```python
# 3. Age Group Risk Analysis
df['AGE'] = df['DAYS_BIRTH'].abs() / 365
df['AGE_BIN'] = pd.cut(
    df['AGE'], 
    bins=[20, 30, 40, 50, 60, 70], 
    labels=['20-30', '30-40', '40-50', '50-60', '60+']
)

age_summary = df.groupby('AGE_BIN')['TARGET'].mean() * 100

plt.figure(figsize=(9, 6))
sns.barplot(x=age_summary.index, y=age_summary.values, palette="rocket")
plt.title("Default Rate berdasarkan Kelompok Usia")
plt.ylabel("Persentase Gagal Bayar (%)")
plt.xlabel("Kategori Usia")
plt.show()
```

```python
# 4. Income Type Risk Analysis
plt.figure(figsize=(10, 6))
income_summary = df.groupby('NAME_INCOME_TYPE')['TARGET'].mean().sort_values(ascending=False) * 100

sns.barplot(x=income_summary.values, y=income_summary.index, palette="mako")
plt.axvline(df['TARGET'].mean() * 100, color='black', linestyle='--', alpha=0.6)
plt.title("Analisis Risiko per Kategori Pekerjaan")
plt.xlabel("Default Rate (%)")
plt.ylabel("")
plt.tight_layout()
plt.show()
```
